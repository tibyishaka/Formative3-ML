```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 3])
y = np.array([3, 6])

m = -1
b = 1
alpha = 0.1
iterations = 4

# To store history for plotting
m_history = [m]
b_history = [b]
error_history = []

n = len(x)

print("Iteration |    m     |    b     |   MSE    |   dm    |   db")
print("------------------------------------------------------------")
for i in range(iterations):
    # Predictions
    y_pred = m * x + b

    # Compute MSE
    error = np.mean((y - y_pred)**2)

    # Gradients
    dm = (-2/n) * np.sum(x * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    # Update parameters
    m = m - alpha * dm
    b = b - alpha * db

    # Save history
    m_history.append(m)
    b_history.append(b)
    error_history.append(error)

    # Print intermediate results
    print(f"{i+1:9} | {m:.4f} | {b:.4f} | {error:.4f} | {dm:.2f} | {db:.2f}")

plt.figure(figsize=(10,4))

# Plot m and b
plt.subplot(1,2,1)
plt.plot(m_history, marker='o', label='m (slope)')
plt.plot(b_history, marker='o', label='b (intercept)')
plt.xlabel('Iteration')
plt.ylabel('Value')
plt.title('Parameter Updates')
plt.legend()

# Plot MSE
plt.subplot(1,2,2)
plt.plot(error_history, marker='o', color='red', label='MSE')
plt.xlabel('Iteration')
plt.ylabel('Mean Squared Error')
plt.title('Error Reduction')
plt.legend()

plt.tight_layout()
plt.show()
```
