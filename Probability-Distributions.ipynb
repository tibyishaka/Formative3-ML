# Select two features

```python
# selecting two feature from the dataset and converting them into an array 
data = df[["Gross_Tertiary_Education_Enrollment", "Gross_Primary_Education_Enrollment"]].values
print(data)
print(data.shape)
```

# Step 2: Compute Mean Vector

```python
#computing the mean vector for your two selected variables
mu = np.mean(data, axis=0)
print(mu)
```

# Step 3: Compute Covariance Matrix Manually

```python
n = data.shape[0]
centered = data - mu
cov_matrix = (centered.T @ centered) / (n - 1)
print("centered")
print(centered)
print("covariance matrix")
print(cov_matrix)
```

# Step 4: Bivariate Normal PDF Function

```python
def bivariate_normal_pdf(x, mu, cov):
    size = len(mu)
    # used the numpy library to find determinant 
    det = np.linalg.det(cov)
    # used the numpy also to find the inverse 
    inv = np.linalg.inv(cov)
    norm_const = 1 / (2 * np.pi * np.sqrt(det))
    
    diff = x - mu
    exponent = -0.5 * diff.T @ inv @ diff
    return norm_const * np.exp(exponent)
```

# Step 5: Compute PDF for Each Data Point

```python
pdf_values = np.array([bivariate_normal_pdf(point, mu, cov_matrix) for point in data])

print("First 10 PDF values:")
print(pdf_values[:10])

```

# Step 6: Create Grid for Visualization

```python
x_min, x_max = data[:, 0].min(), data[:, 0].max()
y_min, y_max = data[:, 1].min(), data[:, 1].max()

x = np.linspace(x_min, x_max, 100)
y = np.linspace(y_min, y_max, 100)
X, Y = np.meshgrid(x, y)

Z = np.zeros(X.shape)

for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        point = np.array([X[i, j], Y[i, j]])
        Z[i, j] = bivariate_normal_pdf(point, mu, cov_matrix)
```

# Step 7: Contour Plot

```python
plt.figure(figsize=(8,6))
plt.contourf(X, Y, Z, levels=30)
plt.scatter(data[:,0], data[:,1], c='red', s=10)
plt.title("Bivariate Normal Distribution (Contour Plot)")
plt.xlabel("Gross_Tertiary_Education_Enrollment")
plt.ylabel("Gross_Primary_Education_Enrollment")
plt.colorbar(label="PDF Value")
plt.show()

```

# Step 8: 3D Surface Plot

```python
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z)
ax.set_title("Bivariate Normal Distribution (3D Surface)")
ax.set_xlabel("Gross_Tertiary_Education_Enrollment")
ax.set_ylabel("Gross_Primary_Education_Enrollment")
ax.set_zlabel("PDF")
plt.show()
```
